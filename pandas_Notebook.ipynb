What is Pandas?

Pandas is a Python library for working with labeled, tabular, and time‑series data.

Core ideas:

DataFrame (2‑D table) and Series (1‑D column).
Built on NumPy for fast, vectorized ops.
Common uses:

Read/write: pd.read_csv, df.to_csv
Inspect: df.head(), df.info(), df.describe()
Clean: df.dropna(), df.fillna(), df.drop_duplicates()
Transform: df.groupby(...).agg(...), pd.merge(), df.pivot()
Time series: pd.to_datetime(), df.resample()
Text: df['col'].str methods; Categorical: pd.get_dummies()
When to use:

Great for EDA, cleaning, and transforming datasets that fit in memory.
For datasets larger than RAM, use Dask/SQL or chunking.

1. Loading and Inspecting the Dataset

```python
import pandas as pd
# Read the CSV file into a DataFrame using read csv method
df=pd.read_csv("Pokemon.csv")
# display the first 5 rows of the DataFrame using head function.
print(df.head())
# info method is used to check colums data types and no null values
print(df.info())
# describe method is used to get a statistical summary of the DataFrame(count, mean, std, quartiles)
print(df.describe())
# shape function tells the rows and columns of datset
print(df.shape)
# to_string function is used to display the entire DataFrame without truncation.
print(df.to_string())
```

2: Missing Values in Dataset

```python
print(df.head(5))
# isnull method is used to check the null values in the dataset. false means value is not null. true means value is null.
df.isnull()
# # we can sum the null values to see how much values are missing in a single column
df.isnull().sum()
# dropna function is used to drop the null values but in bigger dataset when you have more missing values and you drop them. it affects the accuracy.
df.dropna() # now you can see the reduction in rows from 800 to 414.
# fillna function is used to fill the misiing values. we usually use mode function to fill the values.
df["Type 2"]=df["Type 2"].fillna(df["Type 2"].mode()[0])
print(df.head(5))
```

4:Boolean Indexing

```python
df = df[(df['Type 2'] == '') & (df['Type 1'] == '')]
print(df)
# use isin operator to filter data
df = df[(df['Type 1'].isin(['Fire'])) & (df['Type 2'].isin(['Flying']))]
print(df)
```

3: Duplicate Values

```python
# duplicate function is used to check the duplicate values in the dataset. it returns a boolean series where true means duplicate value and false means unique value.
print(df.duplicated())# there is no duplicate value in the dataset.
# drop_duplicates function is used to drop the duplicate values in the dataset. it returns a DataFrame with duplicate values removed.
df.drop_duplicates(inplace=True) # there is no duplicate value in the dataset.
print(df.head(5))

```

4: Types conversion


```python
# astype function is used to change the data type of a column. it takes the data type as an argument and returns a DataFrame with the changed data type.
df['Type 2']=df['Type 2'].astype('category')
print(df.info())
# to numeric function is used to convert a column to numeric data type. it takes the column name as an argument and returns a Series with the converted data type. it uses three parameters coerce, errors and downcast.
df['Type 2']=pd.to_numeric(df['Type 2'], errors='coerce') # it will convert the non-numeric values to NaN
print(df.info())
print(df.head().to_string())
# to_datetime function is used to convert a column to datetime data type. it takes the column name as an argument and returns a Series with the converted data type. it uses three parameters coerce, errors and dayfirst.
# there is no date column in this dataset.

```

5: String Cleaning

```python
# strip function is used to remove the leading and trailing spaces from a string. it takes the column name as an argument and returns a Series with the stripped values.
df['Name']=df['Name'].str.strip()
print(df.head(20))
# lower and upper function is used to convert the string values to lower case and upper case respectively.
df['Type 1']=df['Type 1'].str.lower()
print(df.head(20))
df['Type 1']=df['Type 1'].str.upper()
print(df.head(20))
# replace function is used to replace the values in a column.
df['Type 2']=df['Type 2'].replace('Flying','Fly')
print(df.head(20))
# contains function is used to check if a string contains a specific substring.
print(df['Name'].str.contains('Charizard'))
```

6: Columns/Index Operations


```python
# drop column function is used to drop a column from the DataFrame.You should think before inplacing the drop function because it changes the whole dataset.
df.drop('Total', axis=1, inplace=True)
print(df.head(20))
# rename function is used to rename the columns of the DataFrame.
df.rename(columns={'Type 1':'Primary Type', 'Type 2':'Secondary Type'}, inplace=True)
print(df.head(20))
# reset index function is used to reset the index of the DataFrame. it takes the drop parameter as an argument which is used to drop the old index.
df.reset_index(drop=True, inplace=True)
print(df.head(20))
# set index function is used to set a column as the index of the DataFrame. it takes the column name as an argument and returns a DataFrame with the set index.
df.set_index('Name', inplace=True)
print(df.head(20))
```

7: filtering/Selection of Dataset


```python
# filtering the dataset using loc and iloc function. loc function is used to filter the dataset based on the label of the index and columns. it takes the index label and column label as an argument and returns a DataFrame with the filtered values.
print(df.loc['Bulbasaur']) # it will return the row with the index label 'Bulbasaur'
# iloc function is used to filter the dataset based on the position of the index and columns. it takes the index position and column position as an argument and returns a DataFrame with the filtered values.
print(df.iloc[2]) # it will return the third row of the DataFrame

# query function is used to filter the dataset based on a condition. it takes the condition as an argument and returns a DataFrame with the filtered values.
print(df.query('`Primary Type` == "Poison"')) # it will return the rows where the primary type is poison.
```

8: Outliers

```python
# mask function is used to filter the dataset based on a condition.
mask = (df['Attack']>=100) & (df['Attack']<=100)
print(df[mask])
# clip function is used to filter the dataset based on a condition.
df['Attack'] = df['Attack'].clip(lower=60, upper=100)
print(df.head(20))

```
